## Equation for the Terminal Curve (Flip Bucket)

The curve at the end of the spillway chute is often a **flip bucket**, which is designed to deflect the high-velocity flow and throw it a safe distance away from the toe of the dam. Unlike the parabolic ogee crest, the profile of a flip bucket is a **circular arc**.

The design of this curve involves determining an appropriate radius ($r$) and trajectory angle ($\theta$) to ensure the flow is turned effectively without causing excessive negative pressures on the bucket surface.

*(Figure 7-2 in the source document provides a visual definition of these parameters)*

### Bucket Geometry Equation

The primary equation defining the shape comes from the geometry of a circular arc. The height of the bucket lip ($h$) above the bucket's lowest point (invert) is determined by the bucket radius ($r$) and the desired final trajectory angle ($\theta$):

$$ h = r - r \cos \theta \quad \tag{7-5} $$

where:
- $h$ = Vertical height of the bucket lip above the invert (ft)
- $r$ = Radius of the circular bucket (ft)
- $\theta$ = Trajectory angle the bucket lip makes with the horizontal (degrees)

### Minimum Radius Constraint

To prevent excessively low pressures (and potential cavitation) caused by high centrifugal forces as the water flows through the curve, the radius must be sufficiently large. The minimum recommended radius ($r_{\min}$) is calculated as:

$$ r_{\min} = \frac{\rho V_1^2 d_1}{P_T - \gamma d_1} \quad \tag{7-3} $$

where:
- $V_1$ = Velocity of flow entering the bucket (ft/s)
- $d_1$ = Depth of flow entering the bucket (ft)
- $\rho$ = Mass density of water (slugs/ft³)
- $\gamma$ = Specific weight of water (lb/ft³)
- $P_T$ = Allowable theoretical unit load on the bucket invert (lb/ft²)

A general design guide suggests a bucket radius of at least four times the maximum flow depth ($r \ge 4d_1$).

> **Note for ANSYS Modeling:** When creating the spillway geometry, the chute section will transition from a straight or parabolic profile into this circular arc at the end. Equation 7-5 is used to define the endpoint of this arc based on the chosen radius and exit angle. This terminal geometry is critical for correctly simulating the jet trajectory, downstream impact location, and pressures acting on the bucket itself.
